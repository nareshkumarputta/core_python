```python
# shallow copy-inner objects get shared and modifications occurs in both the copies in inner nested object
# In Python, a shallow copy creates a new outer object, but the nested objects inside it are still shared.
# In a shallow copy:
# Modification to the outer object → does not affect the other copy.
# Modification to nested objects → does affect both copies.

cards=[['hari','rakesh'],['naresh','madhu','udhay'],'uma']
print("original copy Before Shallow copy: ",cards)
import copy
shl_cpy=copy.copy(cards)
print("shallowed copy:",shl_cpy)
shl_cpy[0][0]='vigna'
shl_cpy[2]='bindhu' #outer object → does not affect the other copy.
print("shallow modified copy:",shl_cpy)
print("orginal copy After Shallow copy:",cards)
```

```python
# Deep copy-in Python creates a completely independent copy of an object, including all nested objects.
# Therefore, modifications made to one copy do not affect the other copy.
# Deep copy creates a new object and recursively copies all nested objects,so the original and copied objects are completely independent.
# Modifications do not affect the other copy.

branch=[['ece','cse'],['cse','mech'],'civil']
print("Orginal Copy Before deep copy:",branch)
dp_cpy=copy.deepcopy(branch)
print("deep copy:",dp_cpy)
dp_cpy[0][1]='mme'
dp_cpy[1][0]='aids'
dp_cpy[2]='it'
print("deep modified copy:",dp_cpy)
print("original copy after deep copy",branch)
```

                                                    Higher-Order Functions                                                                                
 
A higher-order function is any function that takes another function as an argument OR returns a 
function. Python has four built-in higher-order functions you'll use constantly: 


▸ map() — Apply a function to every element of an iterable


▸ filter() — Keep only elements where a function returns True


▸ reduce() — Combine all elements into a single value (from functools)


▸ sorted() — Sort any iterable using a custom key function 

                                                map() — Transform Every Element                                                                      
                                                    
map(function, iterable) applies function to each element and returns a map object (convert to list to see 
it). 

```python
l=[1,2,3,4,5]

def square(x):
    return x**2
l=[1,2,3,4,5]
x=list(map(square,l))
print(x)

p=list(map(lambda x:x**2,l))
print(p)

l=[2,3,6,8,9,10]
p=list(map(lambda x:x/2,l))
print(p)
```

A decorator in Python is a function that adds extra functionality to another function without changing its original code
How It Works:


Step 1: Original Function


Step 2: Decorator


Step 3: Decorating

A decorator is a function that takes another function as input, adds extra functionality before or after its execution, and returns the modified function without changing the original function's code

Key Points to Remember:


-A decorator is a higher-order function (it accepts a function and returns a function).


-It helps reuse code instead of repeating the same logic.

@decorator_name is shorthand for:

function = decorator_name(function)


Use *args and **kwargs in decorators to support any function signature.

Common uses include:
Logging,
Authentication/Authorization,
Timing function execution,
Input validation,
Caching,
Access control

                                                    Decorators with Arguments                                                                    
 
Sometimes you want to pass configuration arguments to the decorator itself — not to the function being decorated. This requires one extra level of nesting: a decorator factory. 
 
                                                    Decorator Factory Pattern                                                                   
A decorator factory is a function that RETURNS a decorator. You call the factory with your arguments, it gives back a configured decorator, which then wraps your function

```python
import functools
def dec1(args1):
    def dec(func):
        @functools.wraps(func)
        def wrap(*args,**kwargs):
            for i in range(args1):
                func(*args,**kwargs)
        return wrap
    return dec
@dec1(4)
def say_hello():
    print("Hii")
say_hello()
print(say_hello.__name__)
```

 Logging Decorator--logging decorator automatically records when a function is called, what arguments it received, and what it returned — without touching the function's code. 

```python
import functools 
import datetime 
def logger(func): 
    @functools.wraps(func) 
    def wrapper(*args,**kwargs): 
        timestamp = datetime.datetime.now().strftime('%H:%M:%S')
        print(f'[{timestamp}] Calling: {func.__name__}') 
        print(f'Args:{args},Kwargs: {kwargs}') 
        result=func(*args, **kwargs) 
        print(f'Returned: {result}') 
    return wrapper 
 
@logger 
def multiply(a, b): 
    return a * b 
@logger 
def greet(name, greeting='Hello'): 
    return f'{greeting}, {name}!' 
greet('naresh')
print()
multiply(4, 7) 
```

 Authentication Decorator--Authentication decorators check whether a user has the right to run a function before allowing execution. This pattern is used in every web framework. 

```python
import functools 
def require_role(required_role): 
    def decorator(func): 
        @functools.wraps(func) 
        def wrapper(user, *args, **kwargs): 
            if user.get('role') != required_role: 
                print(f'Access Denied: {user["name"]} is not a {required_role}') 
                return
            print(f'Access Granted: {user["name"]}') 
            return func(user, *args, **kwargs) 
        return wrapper 
    return decorator 
 
@require_role('admin') 
def delete_database(user): 
    print(f'{user["name"]} deleted the database.') 
 
@require_role('student') 
def view_notes(user): 
    print(f'{user["name"]} is viewing notes.') 
 
admin = {'name': 'Alice', 'role': 'admin'} 
guest = {'name': 'Bob',   'role': 'guest'} 
student = {'name': 'Carol', 'role': 'student'} 
 
delete_database(admin) 
# Access Granted: Alice 
# Alice deleted the database. 
delete_database(guest) 
# Access Denied: Bob is not a admin 
view_notes(student) 
# Access Granted: Carol 
# Carol is viewing notes. 
```

Timing / Performance Decorator--A timing decorator measures how long a function takes to execute. This is essential for performance 
profiling and identifying bottlenecks. 

```python
import functools 
import time 
 
def timer(func): 
    @functools.wraps(func) 
    def wrapper(*args, **kwargs): 
        start = time.perf_counter() 
        result = func(*args, **kwargs) 
        end = time.perf_counter() 
        elapsed = end - start 
        print(f'{func.__name__} took {elapsed:.6f} seconds') 
    return wrapper 
 
@timer 
def slow_function(): 
    time.sleep(0.5) 
    return 'done' 
 
@timer 
def compute_sum(n): 
    return sum(range(n)) 
 
slow_function() 
compute_sum(1000000) 
```

```python
 
# Q1.Write a decorator called validate_positive that checks all positional arguments passed to a function. If any argument is negative,
# print an error message and return None without calling the function. Test it on a function multiply(a, b). 
def validate_positive(func):
    def wrap(*args,**kwargs):
        res=args[0]
        for i in args[1:]:
            if res<0 or i<0:
                print("Error")
                return None
            else:
                func(*args)
    return wrap
@validate_positive
def multiply(a,b):
    print(a*b)
multiply(5,7)
multiply(5,-7)
multiply(10,30)
multiply(-10,30)
print()

# Q2.  Create a decorator factory called repeat(n) that runs the decorated function exactly n times. Then stack it with a logger decorator and apply both to a function.
# Trace the exact order in which the wrappers execute. 
def dec_factory(n):
    def dec(func):
        def wrap(*args,**kwargs):
            for i in range(n):
                func(*args,**kwargs)
        return wrap
    return dec
def logger(func):
    @functools.wraps(func)
    def wrap(*args,**kwargs):
        print("Function Called:",func.__name__)
        print("Arguments are:",args)
        print("keyword Arguments are:",kwargs)
        res=func(*args,**kwargs)
        print("function Finished")
        return res
    return wrap
@dec_factory(2)
@logger
def say_hello(name,age):
    print("Name is:",name)
    print("age is:",age)
say_hello("naresh",23)
say_hello(name="Madhu",age=23)
```
