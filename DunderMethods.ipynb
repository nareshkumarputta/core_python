```python
class Student:
    def __init__(self, name, age, marks, subjects):
        self.name = name
        self.age= age
        self.marks = marks
        self.subjects = subjects
    def __str__(self):
        # return f"{self.name} has gotten {self.marks} marks."
        return f"Hey! This is {self.name}!!"
    def __repr__(self):
        return f"Student({self.name}, {self.age}, {self.marks})"
    def __len__(self):
        # return len(self.subjects)
        return len(self.name)
    def __contains__(self, item):
        return item in self.subjects
    def __add__(self, other):
        return self.age + other.age
    def __sub__(self, other):
        return self.marks - other.marks
    def __mul__(self, other):
        return self.age * other.age
    def __truediv__(self, other):
        return len(self.subjects)/len(other.subjects)
    def __floordiv__(self, other):
        return len(self.subjects)//len(other.subjects)
    def __lt__(self, other):
        return self.marks < other.marks
    def __gt__(self, other):
        return self.marks > other.marks
    def __le__(self, other):
        return self.marks <= other.marks
    def __ge__(self, other):
        return self.marks >= other.marks
    def __eq__(self, other):
        return self.marks == other.marks
    def __ne__(self, other):
        return self.marks != other.marks
    def __hash__(self):
        return self.age
obj1 = Student("Alice", 27, 78, ['Maths', 'Science'])
obj2 = Student("Reene", 22, 79, ['Maths', 'Science', 'Geography', 'Civics'])
obj3 = Student("Meg", 25, 72, ['Maths'])
obj4 = Student("Candice", 24, 98, ['Maths', 'Civics', 'Science'])
obj5 = Student("Max", 23, 88, ['English', 'Python'])
print(obj1)
print(obj2)
print(obj4)
print(repr(obj1))
print(len(obj1))
print('English' in obj1)
print('Maths' in obj1)
print(obj1 + obj2)
print(obj3 - obj4)
print(obj2 * obj3)
print(obj1 / obj2)
print(type(obj1 // obj2))
print(obj1 < obj2)
print(obj2 > obj1)
print(obj2 >= obj1)
print(obj5 <= obj3)
print(obj3 == obj2)
print(obj4 != obj5)
print(hash(obj1))
```

```python
# 1. Create a class called Product with attributes name and price. Change the + operator so that adding two Product objects returns the sum of their prices.
# p1 = Product("Keyboard", 1500)
# p2 = Product("Mouse", 700)
# print(p1 + p2)
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def __add__(self,other):
        return self.price+other.price
p1 = Product("Keyboard", 1500)
p2 = Product("Mouse", 700)
print(p1 + p2)
# Expected Output:
# 2200
# Operator to change: +
# Magic Method: __add__()

# 2. Create a class called BankAccount with an attribute balance. Change the - operator so that subtracting one account from another returns the difference between their balances.
# account1 = BankAccount(10000)
# account2 = BankAccount(3500)
# print(account1 - account2)
class bankaccount:
    def __init__(self,balance):
        self.balance=balance
    def __sub__(self,other):
        return self.balance-other.balance
account1 = bankaccount(10000)
account2 = bankaccount(3500)
print(account1 - account2)
# # Expected Output
# 6500
# Operator to change: -
# Magic Method: __sub__()

# 3. Create a class called ShoppingCart with an attribute total. Change the * operator so that multiplying a cart by an integer returns the total cost for that many identical carts.
# cart = ShoppingCart(2500)
# print(cart * 3)
class shoppingcart:
    def __init__(self,total):
        self.total=total
    def __mul__(self, other):
        return self.total*other
cart = shoppingcart(2500)
print(cart * 3)
# Expected Output:
# 7500
# Operator to change: *
# Magic Method: __mul__()

# 4. Create a class called Bill with an attribute amount. Change the / operator so that dividing a Bill object by a number returns the amount each person must pay.
# bill = Bill(1200)
# print(bill / 4)
class bill:
    def __init__(self,amount):
        self.amount=amount
    def __truediv__(self, other):
        return self.amount/other
bill=bill(1200)
print(bill/4)
# Expected Output:
# 300.0
# Operator to change: /
# Magic Method: __truediv__()

# 5. Create a class called Student with attributes name and marks. Change the > operator so that one student is considered greater than another when their marks are higher.
# student1 = Student("Anil", 85)
# student2 = Student("Ravi", 72)
# print(student1 > student2)
class student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def __gt__(self, other):
        return self.marks>other.marks
student1 = student("Anil", 85)
student2 = student("Ravi", 72)
print(student1 > student2)
# Expected Output:
# True
# Operator to change: >
# Magic Method: __gt__()

# 6. Create a class called Book with attributes title and isbn. Change the == operator so that two books are considered equal when their ISBN values are the same, even if their titles are different.
# book1 = Book("Python Basics", "ISBN101")
# book2 = Book("Learn Python", "ISBN101")
# print(book1 == book2)
class book:
    def __init__(self,title,isbn):
        self.title=title
        self.isbn=isbn
    def __eq__(self, other):
        return self.isbn==other.isbn
book1 = book("Python Basics", "ISBN101")
book2 = book("Learn Python", "ISBN101")
print(book1 == book2)
# Operator to change: ==
# Magic Method: __eq__()
# Expected Output:
# True

# 7. Create a class called Score with an attribute points. Change the addition behaviour so that an integer can be added on the left side of a Score object.
# score = Score(80)
# print(20 + score)
class score:
    def __init__(self,points):
        self.points=points
    def __radd__(self,other):
        return other+self.points
core = score(80)
print(20+core)
# Expected Output:
# 100
# Operator to change: Reflected +
# Magic Method: __radd__()

# 8. Create a class called Inventory with an attribute quantity. Change the += operator so that the supplied quantity is added directly to the existing object.
# stock = Inventory(50)
# stock += 20
# print(stock.quantity)
class inventory:
    def __init__(self,quantity):
        self.quantity=quantity
    def __iadd__(self, other):
        self.quantity+=other
        return self
stock = inventory(50)
stock+=20
print(stock.quantity)
# Expected Output:
# 70
# Operator to change: +=
# Magic Method: __iadd__()

# 9. Create a class called Permission with an attribute containing a set of permission names. Change the | operator so that combining two Permission objects returns all unique permissions from both objects.
# p1 = Permission({"read", "write"})
# p2 = Permission({"write", "delete"})
# result = p1 | p2
# print(result.permissions)
class permission:
    def __init__(self,permission_names):
        self.permission_names=permission_names
    def __or__(self, other):
        return self.permission_names | other.permission_names
p1= permission({"read", "write"})
p2 = permission({"write", "delete"})
result = p1 | p2
print(result) 
# Expected Output:
# {'read', 'write', 'delete'}
# Operator to change: |
# Magic Method: __or__()

# 10. Create a class called Team with attributes name and points. Change the <= operator so that one team is considered less than or equal to another when its points are lower than or equal to the other team's points.
# team1 = Team("Falcons", 25)
# team2 = Team("Tigers", 30)
# print(team1 <= team2)
class team:
    def __init__(self,name,points):
        self.name=name
        self.points=points
    def __le__(self,other):
        return self.points<=other.points
team1 = team("Falcons", 25)
team2 = team("Tigers", 30)
print(team1 <= team2)
# Expected Output:
# True
# Operator to change: <=
# Magic Method: __le__()
```
