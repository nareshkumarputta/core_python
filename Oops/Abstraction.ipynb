                                                          ABSTRACTION                                                                                     
Abstraction means hiding unnecessary implementation details and showing only the important functionality to the user.

Abstraction is an OOP concept that hides implementation details and exposes only the essential functionality to the user. In Python, abstraction can be achieved using abstract classes and abstract methods from the abc module.

Real-life example 🚗

When you drive a car:

You use steering → to turn
You use brake → to stop
You use accelerator → to move faster

But you don't need to know how the engine internally works.

That's abstraction:

Show what to do, hide how it is done.



Python provides abstraction mainly through:

from abc import ABC, abstractmethod

We create:

Abstract class
Abstract method

````python
from abc import ABC, abstractmethod
# Very important concept-Look at this:
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass
# Here:
# Vehicle → abstract class
# start() → abstract method
# @abstractmethod → says that child classes must implement this method
# The purpose of Vehicle is to provide a common structure/rule for its child classes.
# We're saying:
# "Every vehicle should have a start() method, but I don't care how each vehicle starts."
# Then:
class Car(Vehicle):
    def start(self):
        print("Car starts with a key")
# and:
class Bike(Vehicle):
    def start(self):
        print("Bike starts with a button")
# Each child decides how to implement start().
# That's abstraction.
c = Car()
c.start()
b = Bike()
b.start()
# Why do we need an abstract class?
# v = Vehicle()
# Python gives an error because Vehicle contains an abstract method.
# You cannot create an object of an abstract class.
# TypeError: Can't instantiate abstract class Vehicle
````

````python
# 1. How do you import ABC and abstractmethod from the abc module?
from abc import ABC, abstractmethod

# 2. What can an abstract class contain?
print('''An abstract class can contain both abstract and normal members.
In Python, an abstract class can contain:
✅ Abstract methods
✅ Concrete (normal) methods
✅ Constructor (__init__)
✅ Instance variables
✅ Class variables
✅ Properties
✅ Static methods
✅ Class methods''')
print()

# 3. Can an abstract class with unimplemented abstract methods be instantiated? Explain.
# No, an abstract class with unimplemented abstract methods cannot be instantiated.
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
# Here, sound() is an abstract method.
# if we try
# a = Animal()
# We get:
# TypeError: Can't instantiate abstract class Animal
# Why?
# Because Animal says:
# "Every child class must provide an implementation for sound()."
# So Python does not allow us to create an object until all abstract methods are implemented.
# Correct approach:
class Dog(Animal):
    def sound(self):
        print("Dog barks")
d = Dog()
d.sound()
print()
# No. An abstract class containing unimplemented abstract methods cannot be instantiated. A concrete child class must implement all 
# abstract methods before its object can be created.
# Easy rule:
# Abstract method present → object cannot be created 

# 4. What must a child class do with all inherited abstract methods to become a concrete class?
# A child class must implement all inherited abstract methods to become a concrete class. 
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
    @abstractmethod
    def move(self):
        pass
# Here, Animal has 2 abstract methods:
# sound()
# move()
# Now suppose the child class implements both
class Dog(Animal):
    def sound(self):
        print("Dog barks")
    def move(self):
        print("Dog runs")
d = Dog()
# Now Dog is a concrete class, so we can create its object:
d.sound()
d.move()
# if 
# class Dog(Animal):
#     def sound(self):
#         print("Dog barks")
# d = Dog()
# TypeError: Can't instantiate abstract class Dog
print('''Abstract Parent
      ↓
Child must implement ALL abstract methods
      ↓
All implemented → Concrete class ✅
Some still unimplemented → Abstract class ❌''')
````

````python
# 5. Can abstractness continue through multilevel inheritance? Explain.
# Yes Abstractness can continue through multilevel inheritance.
# If a child class does not implement all inherited abstract methods, that child class also remains abstract.
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
# Animal is abstract because sound() is abstract.
# Now create another class:
class Mammal(Animal):
    pass
# Mammal does not implement sound(), so Mammal is also abstract.
# Now create another child:
class Dog(Mammal):
    def sound(self):
        print("Dog barks")
# Now Dog implements sound(), so Dog becomes a concrete class.
d = Dog()
d.sound()
# Another possibility
# If Mammal implements sound():
class Mammal(Animal):
    def sound(self):
        print("Some mammal sound")
# Then Mammal becomes concrete, and Dog can inherit that implementation.


# 6. Does Python automatically enforce an exact abstract-method signature? Explain.
# No. Python does not automatically enforce an exact signature for abstract methods.
# An abstract method only requires the child class to implement a method with the same name. Python does not check that the 
# parameters match exactly.
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self, name):
        pass
class Dog(Animal):
    def sound(self):   # Different signature-Python does not give an error just because sound() in Dog doesn't have the name parameter.
        print("Bark")
d = Dog()
d.sound()
# But this is a problem if you call it like the parent method
# d.sound("Tommy")
# TypeError: Dog.sound() takes 1 positional argument but 2 were given
````

````python
# 7. What is the effect of changing parameter names and changing the required parameter count in an abstract method?
# 1. Changing only parameter names
# Changing the parameter names does not prevent the child class from becoming concrete.
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self, name):
        pass
class Dog(Animal):
    def sound(self, animal_name):   # name → animal_name
        print(animal_name, "Barks ")
# This is completely valid.
d = Dog()
d.sound("Tommy")
# Changing parameter names → No effect on abstractness
# 2. Changing the required parameter count
# Python also does not prevent you from changing the number of parameters.
class Animal(ABC):
    @abstractmethod
    def sound(self, name):
        pass
class Dog(Animal):
    def sound(self,name,age):   # extra parameter added
        print("My",name, "sounds Bark")
        print("Age is",age)
# Dog is still considered a concrete class, so this works:
d = Dog()
d.sound('snoopy',12)
````

````python
# 8. Can super() be used to call an abstract parent's constructor or normal method? Explain.
# Yes. super() can be used to call an abstract parent's constructor or a normal method.
# The important point is that abstractness applies to abstract methods, not automatically to the constructor or normal methods.
# 1. Calling the abstract parent's constructor
from abc import ABC, abstractmethod
class Animal(ABC):
    def __init__(self, name):
        self.name = name
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)
    def sound(self):
        print("Bark")
d = Dog("Tommy")
print(d.name)

# 2. Calling a normal method of the abstract parent
class Animal(ABC):
    def show(self):
        print("This is Animal")
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):
    def sound(self):
        super().show()
        print("Bark")
d = Dog()
d.sound()
````

````python
# 9. Why are abstract classes useful for enforcing a common design while supporting polymorphism?
# Abstract classes are useful because they define a common structure that all child classes must follow,
# while allowing each child class to provide its own implementation.
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):
    def sound(self):
        print("Bark")
class Cat(Animal):
    def sound(self):
        print("Meow")
# Now we can use polymorphism:
animals = [Dog(), Cat()]
for animal in animals:
    animal.sound()

# Why is this useful?
# 1. Common design-The abstract class says: Every Animal must have a sound() method.
# So child classes cannot become concrete without implementing it.
# 2. Different implementations
# Dog implements sound() as "Bark" while Cat implements it as "Meow".
# 3. Polymorphism
# We can treat both objects as Animal objects and call the same method:
# animal.sound()
# Python automatically executes the appropriate child's version.
# Simple way to remember
# Abstraction defines what must be done.
# Child classes define how it is done.
# Polymorphism allows us to use them through the same interface.
# So abstract classes provide common design + enforcement, while polymorphism provides different behavior through the same method call.
````

````python
# 1. Payment System 
# Create an abstract class Payment with an abstract method pay(amount)
# Create:
# 	UPI
# 	CreditCard
# 	Cash
# Each class should implement pay() differently.
from abc import ABC,abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay_amount(self):
        pass
class UPI(Payment):
    def pay_amount(self):
        print("payment successfull using UPI")
class CreditCard(Payment):
    def pay_amount(self):
        print("payment successfull using Creditcard")
class Cash(Payment):
    def pay_amount(self):
        print("payment successfull using Cash")
u=UPI()
cc=CreditCard()
c=Cash()
l=[u,cc,c]
for i in l:
    i.pay_amount()
print()

# 2. Notification System 
# Create an abstract class Notification with:
# send(message)
# Create:
# 	Email 
# 	SMS 
# 	WhatsApp 
# Each child class should implement send() differently.
class Notification(ABC):
    @abstractmethod
    def send(self,message):
        return message
class email(Notification):
    def send(self,message,name):
        print("hi",name, message)
class sms(Notification):
    def send(self):
        print("hello user,your account has been blocked")
class whatsapp(Notification):
    def send(self,name,name2):
        print(name ,'sent message to',name2)
e=email()
e.send(' good morning','naresh')
s=sms()
s.send()
w=whatsapp()
w.send('naresh','vigna')
print()

````

````python
# 3. Login System
# Create an abstract class LoginSystem with:
# login(username, password)
# logout()
# Create:
# 	AdminLogin 
# 	UserLogin 
# Implement login differently for admin and normal users.
from abc import ABC, abstractmethod
class LoginSystem(ABC):
    @abstractmethod
    def login(self, username, password):
        pass
    def logout(self):
        print("Logout Successful...!!!")
class AdminLogin(LoginSystem):
    def login(self, username, password):
        print("Login Successful For Admin:", username)
    def logout(self):
        return super().logout()
class UserLogin(LoginSystem):
    def login(self, username, password):
        print("Login Successful For User:", username)
    def logout(self):
        return super().logout()
# Admin Login
a = AdminLogin()
a.login("naresh", "admin123")
a.logout()
# User Login
u = UserLogin()
u.login("akhil", "data analyst")
u.logout()
````

````python
# 4. Bank Account
# Create an abstract class BankAccount with two abstract methods:
# deposit(amount)
# withdraw(amount)
# Create:
# 	SavingsAccount 
# 	CurrentAccount 
# Maintain the balance using an instance variable.
class BankAccount(ABC):
    def __init__(self,balance):
        self.balance=balance
    @abstractmethod
    def deposit(self,amount):
        pass
    @abstractmethod
    def withdraw(self,amount):
        pass
class SavingsAccount(BankAccount):
    def __init__(self,balance):
        super().__init__(balance)
    def deposit(self, amount):
        if amount>0:
            self.balance+=amount
            print("Your Savings Acc Has Been Credited With RS.",amount)
        else:
            raise ValueError("Enter A valid Amount")
    def withdraw(self, amount):
        if amount<=self.balance:
            self.balance-=amount
            print("Your Savings Acc Has Been Debited With RS.",amount)
        else:
            raise ValueError("Enter A valid Amount")
class CurrentAccount(BankAccount):
    def __init__(self,balance):
        super().__init__(balance)
    def deposit(self, amount):
        if amount>0:
            self.balance+=amount
            print("Your Current Acc Has Been Credited With RS.",amount)
            print("Thank you...!!!!")
        else:
            raise ValueError("Enter A valid Amount")
    def withdraw(self, amount):
        if amount<=self.balance:
            self.balance-=amount
            print("Your Current Acc Has Been Debited With RS.",amount)
        else:
            raise ValueError("Enter A valid Amount")
sa=SavingsAccount(20000)
sa.deposit(1000)
sa.withdraw(100)
ca=CurrentAccount(1000000)
ca.deposit(23000)
ca.withdraw(10000)
````

````python
from abc import ABC, abstractmethod
class DataStorage(ABC):
    @property
    @abstractmethod
    def storage_type(self):
        pass
    @abstractmethod
    def save(self, data):
        pass
    @abstractmethod
    def load(self, key):
        pass
    @abstractmethod
    def delete(self, key):
        pass
# --------------------------------
# File Storage
# --------------------------------
class FileStorage(DataStorage):
    @property
    def storage_type(self):
        return "File Storage"
    def save(self, data):
        with open("data.txt", "w") as file:
            file.write(data)
        print("Data saved to file")
    def load(self, key):
        with open("data.txt", "r") as file:
            data = file.read()
        print("Data loaded from file:", data)
        return data
    def delete(self, key):
        open("data.txt", "w").close()
        print("Data deleted from file")
# --------------------------------
# Memory Storage
# --------------------------------
class MemoryStorage(DataStorage):
    def __init__(self):
        self.data = {}
    @property
    def storage_type(self):
        return "Memory Storage"
    def save(self, data):
        self.data[data["key"]] = data["value"]
        print("Data saved to memory")
    def load(self, key):
        value = self.data.get(key)
        print("Data loaded from memory:", value)
        return value
    def delete(self, key):
        self.data.pop(key, None)
        print("Data deleted from memory")
# --------------------------------
# FileStorage object
# --------------------------------
f = FileStorage()
print(f.storage_type)
f.save("Hello Naresh")
f.load("101")
f.delete("101")
# --------------------------------
# MemoryStorage object
# --------------------------------
m = MemoryStorage()
print(m.storage_type)
m.save({
    "key": "101",
    "value": "Hello Naresh"
})
m.load("101")
m.delete("101")

### Why are they interchangeable?

# Both classes have the same methods:

# ```text
# storage_type
# save()
# load()
# delete()
# ```

# But their implementations are different:

# ```text
# FileStorage
#     ↓
# actual data.txt file

# MemoryStorage
#     ↓
# Python dictionary
# ```

# So you can write one function:

def use_storage(storage):
    print(storage.storage_type)
    storage.save("Hello")

# use_storage(FileStorage())
# use_storage(MemoryStorage())

# The **interface is the same**, but the **implementation is different**. That's the main idea of abstraction + polymorphism.
````
