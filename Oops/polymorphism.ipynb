                                                        Polymorphism                                                                              

One name, many forms.

In programming, the same method/function/operator can behave differently depending on the object or data given to it.

```python
print(10 + 20)
print("Hello " + "Naresh")
# The + operator is the same, but it behaves differently:
# With numbers → addition
# With strings → concatenation
# That's polymorphism.
```

```python
class Animal:
    def sound(self):
        print("Animal sound")
# Then many child classes:
class Dog(Animal):
    def sound(self):
        print("Bark")
class Cat(Animal):
    def sound(self):
        print("Meow")
class Cow(Animal):
    def sound(self):
        print("Moo")
animals = [Dog(), Cat(), Cow()]
for animal in animals:
    animal.sound()

# Output:
# Bark
# Meow
# Moo
# You don't need:
# if animal == Dog:
#     ...
# elif animal == Cat:
#     ...
# elif animal == Cow:
#     ...
# Python automatically calls the appropriate sound().
# This is a major benefit of polymorphism.
```

                                                     Types of Polymorphism                                                                        

1.Duck Typing--If an object behaves like the required type, we don't care what its actual type is.

2.Method Overriding--Method overriding happens when A child class provides its own implementation of a method that already exists in the parent class.

3.Operator Overloading--Operators can behave differently depending on the operands.

4.Method Overloading — Python doesn't support traditional overloading directly, but we can achieve similar behavior using default arguments, *args, etc.

```python
# 6. Duck Typing
# Create three completely unrelated classes: PDFReport, ExcelSheet, and EmailMessage. Each has a method send(). 
# Write a function dispatch(item) that calls send() using duck typing. Prove that no inheritance is needed.
class pdfreport:
    def send(self):
        print("sent succesfull-pdfreport")
class excelsheet:
    def send(self):
        print("sent succesfull-excelsheet")
class emailmessage:
    def send(self):
        print("sent succesfull-emailmesage")
a=pdfreport()
b=excelsheet()
c=emailmessage()
l=[a,b,c]
def dispatch(item):
    for i in item:
        i.send()
dispatch(l)
print()
```

```python
# 5.Method Overriding
# Build a payment system with a base class Payment(amount) and three subclasses: CreditCard, UPI, and NetBanking. Each overrides a
# method process() with its own logic. Write a function checkout(payment) that calls process() on any payment object and demonstrate polymorphism.
class payment:
    def __init__(self,amount):
        self.amount=amount
    def process(self):
        print("Process Succesfull by payment")
class creditcard(payment):
    def process(self):
        print("Process Succesfull by creditcard")
class upi(payment):
    def process(self):
        print("Process Succesfull by upi")
    pass
class netbanking(payment):
    def process(self):
        print("Process Succesfull by netbanking")
a=creditcard(1800)
b=upi(9000)
c=netbanking(4000)
l=[a,b,c]
def checkout(payment):
    for i in payment:
        i.process()   
        #All three objects use the same method name
        #But they produce different results
        #That's polymorphism
checkout(l)
```

```python
# 7. Polymorphism + Duck Typing
# Write a function total_area(shapes) that takes a list of any shape objects and returns their combined area. 
# Use duck typing — any object with an area() method should work.
# Test with Circle, Rectangle, Triangle, and a custom class Hexagon.
class circle:
    def __init__(self,radius):
        self.radius=radius
    def method(self):
        self.area= 3.14*self.radius*self.radius
        return self.area
class rectangle:
    def __init__(self,l,b):
        self.l=l
        self.b=b
    def method(self):
        self.area=self.l*self.b
        return self.area
class triangle:
    def __init__(self,b,h):
        self.b=b
        self.h=h
    def method(self):
        self.area=0.5*self.b*self.h
        return self.area
class hexagon:
    def __init__(self,side):
       self.side=side
    def method(self):
        self.area=2.6*self.side
        return self.area
c=circle(56)
r=rectangle(4,8)
t=triangle(3,5)
h=hexagon(34)
shapes=[c,r,t,h]
def total_area(shapes):
    area=0
    for i in shapes:
        area+=i.method()
    return area 
print(total_area(shapes))
print()      
```

```python
# 8. Concept Question
# Explain the difference between method overriding and method overloading. Python does not natively support overloading — 
# show how to simulate it using *args or default parameters.
class a:
    def add(self,a,b):
        return a+b
    def add(self,a,b,c): #The second add() replaces the first one.
        return a+b+c
d=a()
print(d.add(10,50,80))
print()
# e=a()
# print(e.add(10,20)) would produce an error because the final add() expects 3 arguments.

# How Python Achieves Similar Behavior
# We can use default arguments.
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c
c = Calculator()
print(c.add(10, 20))
print(c.add(10, 20, 30))

# or by using *args
class b:
    def add(self,*args):
        sum=0
        for i in args:
            sum+=i
        return sum
c=b()
print(c.add(10,20,30,50))
```

```python
# operator overloading
# 6. Creating Our Own Operator Behavior
# Python allows us to define how operators behave for our classes.
class Student:
    def __init__(self, marks):
        self.marks = marks
    def __add__(self, other):
        return self.marks - other.marks
s1 = Student(80)
s2 = Student(90)
print(s1 + s2)
# Magical Methods
# __init__()
# __str__()
# __add__()
# __sub__()
# __mul__()
# __eq__()
# __lt__()
# __gt__()
```
