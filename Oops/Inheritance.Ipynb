```python
# 1. Bank Management System Create a Bank class with: • balance variable • deposit() • withdraw() • check_balance()
#  Create a User class that inherits Bank and displays the user's name. Perform deposit, withdrawal, and balance check. 
class Bank:
    balance=10000
    def __init__(self,name):
        self.name=name
    def deposit(self,amount):
        Bank.balance+=amount
        print("Your Account Has been credited with Amount Rs.",amount)
    def withdraw(self,amount):
        global balance
        Bank.balance-=amount
        print("Your Account Has been debited with Amount Rs.",amount)
    def check_balance(self):
        print("Your Account Balance is Rs.",Bank.balance)
class user(Bank):
    pass
a=user('Naresh')
print(a.name)
a.deposit(1000)
a.withdraw(500)
a.check_balance()
```

```python
# 2. Employee Salary System Create an Employee class with: • emp_name  • salary • display_details()
#  Create a Manager class that inherits Employee and adds a bonus(). Display the total salary. 
class Employee:
    def __init__(self,emp_name,salary):
        self.emp_name=emp_name
        self.salary=salary
    def display_details(self):
        return {"Name of the Employee ":self.emp_name,
         "Salary of the Employee:":self.salary
        }
class Manager(Employee):
    def bonus(self,bonus_percent):
        self.salary=self.salary*bonus_percent+self.salary
        print("Your Total Salary After Adding Bonus :",self.salary)
a=Manager("naresh",1000)
print(a.display_details())
a.bonus(0.15)
```

```python
# 3. Student Result System Create a Student class with: • Name  • marks • display_marks()
#  Create a Result class that inherits Student and calculates whether the student has passed or failed. 
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def display_marks(self):
        print({'name':self.name,
               'marks':self.marks})
class Results(Student):
    def calc(self):
        if self.marks>=45:
            print("Passed")
        else:
            print("Failed")
a=Results('naresh',50)
a.display_marks()
a.calc()
b=Results('naresh',34)
b.calc()
```

```python
# 4. Food Ordering System Using Multilevel Inheritance 
# Class 1: Restaurant • Create a method menu(item) that returns the price of the selected food item. 
#  Class 2: FoodCourt (inherits Restaurant) Create the following methods: • display_menu() – Display the available food items. 
#  • order() – Accept the food item from the user and allow multiple orders.  • 
# billing() – Display the total bill and add a packing charge of ₹20. 
#  Class 3: Customer (inherits FoodCourt) • Create an object of the Customer class. 
#  • Call the order() method.  
class Restaurent:
    def menu(self,item):
        if item=='pizza':
            return 250
        elif item=='chicken_biryani':
            return 450
        elif item=='mutton_biryani':
            return 4850
        else:
            return 0
class Foodcourt(Restaurent):
    def __init__(self):
        self.total=0
    def display_menu(self):
        print("Pizza - 250")
        print("Chicken Biryani - 450")
        print("Mutton Biryani - 4850")
    def order(self):
        while True:
            item=input("enter your item ")
            price=self.menu(item)
            if price==0:
                print("Item Not Available")
            else:
                self.total+=price
                print("Added",item,price)
            choice=input(("Do u want to order more (yes/no): "))
            if choice=='no':
                break
    def billing(self):
        print("Food Total:",self.total)
        print("Packing Charges: 20")
        self.total+=20
        print('Total Bill:',self.total)
class Customer(Foodcourt):
    pass
naresh=Customer()
naresh.display_menu()
naresh.order()
naresh.billing()
        
```

```python
# 5. Movie Ticket Booking System Using Multilevel Inheritance Class 1: Movie • Create a method ticket(movie) that returns the ticket price. 
#  Class 2: Booking (inherits Movie) Create the following methods: • movies() – Display the available movies.  
# • selection() – Allow the user to book multiple tickets.  • billing() – Display the total amount and add a booking charge of ₹30. 
#  Class 3: Customer (inherits Booking) • Create an object and call the selection() method.  
class movie:
    def ticket(self,movie):
        if movie=='peddi':
            return 250
        elif movie=='paradise':
            return  200
        elif movie=='kalki':
            return 500
        elif movie=='raaka':
            return 1600
        else:
            return 0
class booking(movie):
    def __init__(self):
        self.total=0
    def display_details(self):
        print("peddi-250\n paradise 200\n kalki 500\n rakaa 1600")
    def selection(self):
        while True:
            movie=input("enter the movie name to book")
            price=self.ticket(movie)
            if price==0:
                print("Movie Not Available")
            else:
                self.total+=price
                print("Ticket Booking Succesfull for",movie,price)
            choice=input("Do you want to book tickets for more movies (yes/no)")
            if choice=='no':
                break
    def billing(self):
        print("Total Price :",self.total)
        print("Booking charges Rs.30")
        print("Final Charges :",self.total+30)
class Customer(booking):
    pass
n=Customer()
n.display_details()
n.selection()
n.billing()
```

```python
# 6. Online Course Enrollment System Using Multilevel Inheritance Class 1: Course • Create a method fee(course) that returns the course fee.  
# Class 2: Academy (inherits Course) Create the following methods: • courses() – Display available courses. 
#  • enroll() – Allow the user to enroll in multiple courses.  • billing() – Display the total fee and add a registration fee of ₹100. 
#  Class 3: Student (inherits Academy) • Create an object and call the enroll() method.  
class course:
    def fee(self,course):
        if course=='da':
            return 4500
        elif course=='ai':
            return 35000
        elif course=='ai/ml':
            return 45000
        else:
            return 0
class academy(course):
    def __init__(self):
        self.total=0
    def courses(self):
        print('da-4500')
        print('ai-35000')
        print('ai/ml-45000')
    def enroll(self):
        while True:
            cour=input("Enter the course you want to register")
            price=self.fee(cour)
            if price==0:
                print("Course not avaialable")
            else:
                self.total+=price
                print("You Have Successfully Registered for the",cour,'of Rs.',price)
            choice=input("Do you want to register more courses (yes/no)")
            if choice=='no':
                break
    def billing(self):
            print("Total Fee :",self.total)
            print("registration fee of ₹100.")
            print("Final Charges :",self.total+100)
class student(academy):
    pass
n=student()
n.courses()
n.enroll()
n.billing()
```

```python
# 7. Cab Booking System Using Hierarchical Inheritance Class 1: Cab • Create methods to calculate the fare for Bike, Auto, and Car rides. 
#  Class 2: Uber (inherits Cab) • Create the methods menu(), booking(), and billing().  
# • Add 10% GST and apply a 15% discount if the bill is above ₹1000. 
#  Class 3: Ola (inherits Cab) • Create the methods menu(), booking(), and billing(). 
#  • Add 12% GST and apply a 20% discount if the bill is above ₹1500.  Driver Code •
#  Ask the user to choose Uber or Ola and call the booking() method
class cab:
    def fare(self,book):
        if book=='bike':
            return 10
        elif book=='auto':
            return 15
        elif book=='car':
            return  30
class uber(cab):
    def __init__(self):
        self.price=0
    def menu(self):
        print("1.Bike - ₹10/km\n2. Auto - ₹15/km \n3. car - ₹20/km")
    def booking(self):
        book=input("enter the cab you want to book")
        distance=int(input("enter the distance you want to travel"))
        self.price=self.fare(book)*distance
    def billing(self):
        print("Total Fare:",self.price)
        gst=0.10*self.price
        self.price+=gst
        print("Added Gst of 10%")
        # print("Final Charge after Gst Added:",self.price+gst)
        if self.price>1000:
            self.price=self.price*0.15-self.price
            print("Your Final Fare is Rs.",self.price)
        else:
            print("Your Final Fare is Rs.",self.price)
class ola(cab):
    def __init__(self):
        self.price=0
    def menu(self):
        print("1.Bike - ₹10/km\n2. Auto - ₹15/km \n3. car - ₹20/km")
    def booking(self):
        book=input("enter the cab you want to book")
        distance=int(input("enter the distance you want to travel"))
        self.price=self.fare(book)*distance
    def billing(self):
        print("Total Fare:",self.price)
        gst=0.12*self.price
        self.price+=gst
        print("Added Gst of 12%")
        if self.price>1000:
            self.price=self.price*0.20-self.price
            print("Your Final Fare is Rs.",self.price)
        else:
            print("Your Final Fare is Rs.",self.price)
# p=input("enter your choice")

n=ola()
n.menu()
n.booking()
n.billing()
p=uber()
p.menu()
p.booking()
p.billing()
```
