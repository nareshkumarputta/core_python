                                                          Encapsulation                                                                            

Encapsulation means wrapping data and the methods that operate on that data inside a class, while controlling how that data can be accessed or modified.

In simple words:
Keep data + related functions together, and protect the data from unwanted direct access.

We need some level of access control.

Python provides three naming conventions:

1. Public--A public variable can be accessed from anywhere.
2. Protected--Python uses a single underscore to indicate a protected member.
3. Private--Python uses double underscore:

```python
# Public Members
# A public variable can be accessed from anywhere.

class Student:
    def __init__(self):
        self.name = "Naresh"
            # self.name is public.
            # Rule
            # Normal variable:
            # self.name → Public.
s = Student()
print(s.name)

# we can also modify it:
s.name = "Rahul"
print(s.name)
```

```python
# Protected Members _
# Python uses a single underscore to indicate a protected member.

class Student:
    def __init__(self):
        self._marks = 80
# The _ means: "This is intended for internal/subclass use; please don't access it directly."
# Wait — Python allows this! So is _marks actually protected?

# Important:
# Python does not enforce protected access like some languages.
# It is mainly a convention.
# _marks means: "Developers, please treat this as protected."
s = Student()
print(s._marks)
# print(s.marks)
```

```python
# Protected with Inheritance
class Parent:
    def __init__(self):
        self._value = 100
class Child(Parent):
    def show(self):
        print(self._value)
c = Child()
c.show()
```

```python
# Private Member
class BankAccount:
    def __init__(self):
        self.__balance = 10000
account = BankAccount()
# print(account.__balance) because __balance is not directly accessible using that name from outside the class.
```

Getter and Setter

When a variable is private, we often use:

Getter-To read the value.


Setter-To modify the value.

```python
class Student:
    def __init__(self, marks):
        self.__marks = marks
    def get_marks(self):
        return self.__marks
    def set_marks(self, marks):
        self.__marks = marks
s = Student(80)
print(s.get_marks())
s.set_marks(90)
print(s.get_marks())
```

Why Setter is Useful

Suppose marks should always be between 0 and 100.

```python
class Student:
    def __init__(self, marks):
        self.__marks = marks
    def get_marks(self):
        return self.__marks
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks")
s = Student(80)
s.set_marks(90)
print(s.get_marks())
s.set_marks(150)
# This is a major advantage of encapsulation.
```

```python
# Private Doesn't Mean Completely Invisible-This is an important Python interview question.

# You may think:
# self.__balance-is completely inaccessible.
# Not exactly.Python uses something called name mangling.
class Bank:
    def __init__(self):
        self.__balance = 10000
b = Bank()
print(b.__dict__)
print(b._Bank__balance)
# Python internally changes:
# __balance-to something like:_Bank__balance
# This is called name mangling.
# So technically, it can still be accessed.
# But you shouldn't normally do this.
# Therefore, remember:
# Python's __private members provide name mangling, not absolute security.
```

```python
class Bank:
    bank_name = "ABC Bank"
    def __init__(self,balance):
        self.__balance = balance
    def getbalance(self):
        return self.__balance
    def setbalance(self, new_balance):
        self.__balance = new_balance
alice = Bank(10000)
# print(alice.__balance)  --> error
print(alice._Bank__balance) #-->access balance using name mangling
print(alice.getbalance())
alice.setbalance(15000)
print(alice.getbalance())
```

                                                     Encapsulation with @property                                                                

Python provides a cleaner way to implement getters and setters using:

@property

```python
class Student:
    def __init__(self, marks):
        self.__marks = marks
    @property
    def marks(self):
        return self.__marks
    @marks.setter
    def marks(self, value):
        if 0 <= value <= 100:
            self.__marks = value
        else:
            print("Invalid marks")
s = Student(80)
print(s.marks)
s.marks = 90
print(s.marks)

# Notice we don't write:
# s.get_marks()
# Instead:
# s.marks
# And setter:
# s.marks = 90
# This makes the class easier to use.

# Why @property is useful
# Compare:
# Traditional getter/setter
# s.get_marks()
# s.set_marks(90)
# Property
# print(s.marks)
# s.marks = 90
# The second approach looks like a normal variable but still allows us to control access.
```

```python
class Bank:
    bank_name = "ABC Bank"
    def __init__(self,balance):
        self.__balance = balance
    @property
    def balance(self):
        user = input("enter username")
        pin = int(input("enter pin"))
        if user == "naresh" and pin == 4334:
            return self.__balance
        else:
            return "Access Denied!!"
    @balance.setter
    def balance(self, new_balance):
        if new_balance > 0 :
            self.__balance = new_balance
        else:
            print("Balance cannot be less than 0!!!")
alice = Bank(10000)
print(alice.balance)
alice.balance = -1000
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary
    def get_salary(self):
        return self.__salary
    def set_salary(self, salary):
        if salary > 0:
            self.__salary = salary
        else:
            print("Invalid salary")
    def work(self):
        print("Employee works")

class Developer(Employee):
    def work(self):
        print("Developer writes code")
class Tester(Employee):
    def work(self):
        print("Tester tests software")
d = Developer("Naresh", 50000)
t = Tester("Rahul", 40000)
# Encapsulation:
d.set_salary(55000)
print(d.get_salary())
# Polymorphism:
employees = [d, t]
for emp in employees:
    emp.work()
```

```python
# Encapsulation
# 9. @property + @setter
# Design a BankAccount class with a private __balance. Create @property balance as a getter. 
# Create a @balance.setter that raises ValueError if the value is negative. Add methods deposit(amount) and withdraw(amount)
#  that update balance through the setter. Test all edge cases.
class bankaccount:
    def __init__(self,balance):
        self.__balance=balance
    @property
    def balance(self):
        return self.__balance
    @balance.setter
    def balance(self,new_bal):
        if new_bal>=0:
            self.__balance=new_bal
        else:
            raise ValueError ("Balance Cant Be Negative")
    def withdraw(self,amount):
        if amount>self.__balance:
            raise ValueError("insufficients funds")
        elif amount<=self.__balance:
            self.__balance-=amount
            print("Dear Customer Your Acc Has Been Debited With Rs.",amount)
        else:
            raise ValueError("Withdrawl amount must be positive")
    def deposit(self,amount):
        if amount>0:
            self.__balance+=amount
            print("Dear Customer Your Acc Has Been credited With Rs.",amount)
        else:
            raise ValueError("Deposied Amount Must Be Positive")
a = bankaccount(10000)
print("Initial Balance:", a.balance)
a.deposit(2000)
print("Balance:", a.balance)
a.withdraw(3000)
print("Balance:", a.balance)
```

```python
# 10. Computed Properties
# Create a class Circle with a private __radius. Add @property radius (getter) and @radius.setter that validates the radius is positive.
#  Add computed @property area and @property circumference. Show that changing radius automatically updates both computed properties.
class circle:
    def __init__(self,radius):
        self.__radius=radius
    @property
    def radius(self):
        return self.__radius
    @radius.setter
    def radius(self,new_radius):
        if new_radius<=0:
            raise ValueError("Radius Cant be Negative")
        else:
            self.__radius=new_radius
    @property
    def area(self):
        return 3.14*self.__radius*self.__radius
    @property
    def circumference(self):
        return self.__radius*2*3.14
c=circle(34)
print(c.radius)
print(c.area)
print(c.circumference)
c.radius=45
print(c.radius)
print(c.area)
print(c.circumference)
```

```python
# def nearest_prime(n):
def nearest_prime(n):
    ap=bp=0
    for i in range(n+1,n+10):
            fc=0
            for j in range(1,i+1):
                if i%j==0:
                    fc+=1
            if fc==2:
                ap=i
                break
    for i in range(n-1,n-10,-1):
        fc=0
        for j in range(1,i+1):
            if i%j==0:
                fc+=1
        if fc==2:
            bp=i
            break
    if (ap-n)>(n-bp):
        print(bp,end=' ')
    elif (ap-n)<(n-bp):
        print(ap,end=' ')
    else:
        print(ap,bp,end=' ')
l=list(map(int,input().split()))
for i in range(len(l)):
    nearest_prime(l[i])
# nearest_prime(7)
```
